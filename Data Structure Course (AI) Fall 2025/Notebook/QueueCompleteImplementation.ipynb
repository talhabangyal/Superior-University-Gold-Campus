```python
class Queue:
    def __init__(self, total_value):
        self.queue = []
        print("Queue: ", self.queue)
        self.limit = total_value
        print("Limit of Queue is: ", self.limit)

    def enqueue(self):
        if len(self.queue) == self.limit:
            print("Queue is already Filled, and there is no space to insert value in it")
        else:
            for i in range(self.limit - len(self.queue)):
                e = input("Enter Value for Queue: ")
                self.queue.append(e)
                print("Updated Queue is: ", self.queue)

    def dequeue(self):
        if not self.queue:
            print("Queue is Already Empty")
        else:
            j = self.queue.pop(0)  # remove from front
            print("Removed value is: ", j)
            print("Updated Queue is: ", self.queue)

    def front(self):
        if not self.queue:
            print("Queue is Empty and there is no Front Value")
        else:
            print("Front value in Queue is: ", self.queue[0])

    def last(self):
        if not self.queue:
            print("Queue is Empty and there is no Last Value")
        else:
            print("Last value in Queue is: ", self.queue[-1])

    def isFull(self):
        if len(self.queue) == self.limit:
            print("Queue is Full")
        else:
            print("Queue is not Full")
            print("Queue is: ", self.queue)

    def isEmpty(self):
        if not self.queue:
            print("Queue is Empty")
        else:
            print("Queue is not Empty")
            print("Queue is: ", self.queue)


# A generic menu function for Queue
def menu(queue_obj):
    while True:
        print("\nPlease Select any Operation:")
        print("1 - Enqueue")
        print("2 - Dequeue")
        print("3 - Front Value")
        print("4 - Last Value")
        print("5 - isEmpty")
        print("6 - isFull")
        print("7 - Exit")

        choice = int(input("Please select option: "))
        if choice == 1:
            queue_obj.enqueue()
        elif choice == 2:
            queue_obj.dequeue()
        elif choice == 3:
            queue_obj.front()
        elif choice == 4:
            queue_obj.last()
        elif choice == 5:
            queue_obj.isEmpty()
        elif choice == 6:
            queue_obj.isFull()
        elif choice == 7:
            break
        else:
            print("Wrong option! Please select option ranging from 1 to 7")


# Example usage with multiple objects
obj1 = Queue(4)
obj2 = Queue(3)

print("\n--- Menu for obj1 ---")
menu(obj1)

print("\n--- Menu for obj2 ---")
menu(obj2)
```

```python
class Queue:
    def __init__(self, limit):
        self.queue = []
        self.limit = limit
        print(f"\n✅ New Queue created with limit {self.limit}")

    def enqueue(self):
        if len(self.queue) == self.limit:
            print("⚠️ Queue is already full! Cannot enqueue more values.")
        else:
            while True:
                try:
                    value = int(input("Enter an integer value to enqueue: "))
                    self.queue.append(value)
                    print(f"✅ Value {value} enqueued! Current Queue: {self.queue}")
                    break
                except ValueError:
                    print("❌ Invalid input! Please enter an INTEGER only.")

    def dequeue(self):
        if not self.queue:
            print("⚠️ Queue is already empty! Nothing to dequeue.")
        else:
            removed = self.queue.pop(0)
            print(f"✅ Removed value (from front): {removed}")
            print("Current Queue:", self.queue)

    def peek(self):
        if not self.queue:
            print("⚠️ No front value (Queue is empty).")
        else:
            print(f"🔝 Front Value: {self.queue[0]}")

    def isFull(self):
        if len(self.queue) == self.limit:
            print("✅ Queue is FULL")
        else:
            print(f"ℹ️ Queue is NOT Full. Current size: {len(self.queue)}/{self.limit}")

    def isEmpty(self):
        if not self.queue:
            print("✅ Queue is EMPTY")
        else:
            print("ℹ️ Queue is NOT Empty. Current Queue:", self.queue)

    def sort(self):
        if not self.queue:
            print("⚠️ Cannot sort an empty queue.")
            return
        ascending = sorted(self.queue)
        descending = sorted(self.queue, reverse=True)
        print("⬆️ Ascending Order:", ascending)
        print("⬇️ Descending Order:", descending)


# Menu function
def menu(queue_obj):
    while True:
        print("\n📌 Select a Queue Operation:")
        print("1 - Enqueue")
        print("2 - Dequeue")
        print("3 - Sort")
        print("4 - Peek (Front Value)")
        print("5 - Check if Empty")
        print("6 - Check if Full")
        print("7 - Exit")

        try:
            choice = int(input("👉 Enter your choice (1-7): "))
        except ValueError:
            print("❌ Invalid input! Please enter a number between 1-7.")
            continue

        if choice == 1:
            queue_obj.enqueue()
        elif choice == 2:
            queue_obj.dequeue()
        elif choice == 3:
            queue_obj.sort()
        elif choice == 4:
            queue_obj.peek()
        elif choice == 5:
            queue_obj.isEmpty()
        elif choice == 6:
            queue_obj.isFull()
        elif choice == 7:
            print("👋 Exiting menu...")
            break
        else:
            print("⚠️ Wrong option! Please select between 1 and 7.")


# Example usage
q1 = Queue(4)
q2 = Queue(3)

print("\n--- Menu for Queue q1 ---")
menu(q1)

print("\n--- Menu for Queue q2 ---")
menu(q2)
```

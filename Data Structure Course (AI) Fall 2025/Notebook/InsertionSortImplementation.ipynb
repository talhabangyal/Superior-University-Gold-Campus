```python
def insertion_sort(arr):
    # Traverse from the second element to the end of the array
    for i in range(1, len(arr)):
        key = arr[i]  # Current element to be inserted at the correct position
        j = i - 1
        
        # Move elements of arr[0..i-1], that are greater than the key, to one position ahead
        while j >= 0 and key < arr[j]:
            arr[j + 1] = arr[j]  # Shift elements to the right
            j -= 1
        
        arr[j + 1] = key  # Insert the key at the correct position

# Example usage:
arr = [12, 11, 13, 5, 6]
print("un-Sorted array is:", arr)
insertion_sort(arr)
print("Sorted array is:", arr)
```

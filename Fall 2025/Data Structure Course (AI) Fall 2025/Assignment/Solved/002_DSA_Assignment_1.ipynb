```python
# ==============================
# Task 1: Integrated Document Management System
# ==============================

class Queue:
    def __init__(self, total_value):
        self.queue = []
        self.limit = total_value

    def enqueue(self, value):
        if len(self.queue) == self.limit:
            print("Queue is full")
        else:
            self.queue.append(value)
            print("Updated Queue:", self.queue)

    def dequeue(self):
        if not self.queue:
            print("No jobs to process")
        else:
            job = self.queue.pop(0)
            print("Processed Job:", job)
            print("Updated Queue:", self.queue)

    def peek(self):
        if not self.queue:
            print("No jobs waiting")
        else:
            print("Next job to process:", self.queue[0])


class Stack:
    def __init__(self, total_value):
        self.stack = []
        self.limit = total_value

    def push(self, value):
        if len(self.stack) == self.limit:
            print("Stack is full")
        else:
            self.stack.append(value)
            print("Updated Stack:", self.stack)

    def pop(self):
        if not self.stack:
            print("No revisions to undo")
        else:
            removed = self.stack.pop()
            print("Undone Revision:", removed)
            print("Updated Stack:", self.stack)

    def peek(self):
        if not self.stack:
            print("Revision history is empty")
        else:
            print("Current Revision:", self.stack[-1])

    def show_history(self):
        if not self.stack:
            print("Revision history is empty")
        else:
            print("Revision History:", self.stack)





print("\n==============================")
print(" Task 1: Document Management ")
print("==============================")

s_size = int(input("Enter stack size (for revisions): "))
q_size = int(input("Enter queue size (for print jobs): "))
stack = Stack(s_size)
queue = Queue(q_size)

while True:
    print("\n--- Document Management System ---")
    print("1. Push new document revision")
    print("2. Undo latest revision")
    print("3. Peek current revision")
    print("4. Show full revision history")
    print("5. Enqueue new print job")
    print("6. Peek next print job")
    print("7. Sort and process print jobs")
    print("8. Exit")
    choice = input("Enter choice: ")

    if choice == "1":
        rev = input("Enter revision description: ")
        stack.push(rev)

    elif choice == "2":
        stack.pop()

    elif choice == "3":
        stack.peek()

    elif choice == "4":
        stack.show_history()

    elif choice == "5":
        job = int(input("Enter print job priority: "))
        queue.enqueue(job)

    elif choice == "6":
        queue.peek()

    elif choice == "7":
        if not queue.queue:
            print("No jobs to process")
        else:
            print("\nSorting print jobs using Bubble Sort:")
            bubble_sort(queue.queue)
            print("Final Sorted Queue:", queue.queue)

            print("\nProcessing Jobs:")
            while queue.queue:
                queue.dequeue()

    elif choice == "8":
        print("Program is exiting")
        break

    else:
        print("Invalid choice. Try again.")

def bubble_sort(arr):
    n = len(arr)
    for i in range(n - 1):
        swapped = False
        for j in range(n - i - 1):
            if arr[j] > arr[j + 1]:  # ascending order
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        print(f"After pass {i + 1}:", arr)
        if not swapped:
            break
```

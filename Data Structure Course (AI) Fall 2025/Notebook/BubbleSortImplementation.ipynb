```python
def bubble_sort(arr, show_steps=False):
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                print(arr)
            else:
                print(arr)
        print("")
        if show_steps:
            print(f"After pass {i+1}: {arr}")

# Example usage:
data = [64, 34, 25, 12, 22, 11, 90]
print("Unsorted list:", data)

bubble_sort(data, show_steps=True)

print("Sorted list:", data)
```

```python
class Stack:
    def __init__(self, total_value):
        self.stack = []
        print("Stack is: ", self.stack)
        self.limit = total_value
        print("Limit is: ", self.limit)

    def push(self):
        if len(self.stack) == self.limit:
            print("Stack is already Filled")
        else:
            for i in range(self.limit - len(self.stack)):
                e = input("Enter values for Stack: ")
                self.stack.append(e)
                print("Updated Stack: ", self.stack)

    def pop(self):
        if not self.stack:
            print("Stack is already Empty")
        else:
            j = self.stack.pop()
            print("Removed Value is: ", j)
            print("Updated Stack is: ", self.stack)

    def peak(self):
        if not self.stack:
            print("No top value exist")
        else:
            print("Top Value in Stack is: ", self.stack[-1])

    def isFull(self):
        if self.limit == len(self.stack):
            print("Stack is Full")
        else:
            print("Stack is Not Full")
            print("Stack: ", self.stack)

    def isEmpty(self):
        if not self.stack:
            print("Stack is Empty")
        else:
            print("Stack is Not Empty")
            print("Stack: ", self.stack)

    def sort(self):
        self.stack.sort()
        print("Stack in Ascending Order: ", self.stack)
        self.stack.sort(reverse=True)
        print("Stack in Descending Order: ", self.stack)

    # 🔹 Integrated Bubble Sort for the Stack
    def bubble_sort(self):
        n = len(self.stack)
        for i in range(n):
            swapped = False
            for j in range(0, n - i - 1):
                if self.stack[j] > self.stack[j + 1]:
                    self.stack[j], self.stack[j + 1] = self.stack[j + 1], self.stack[j]
                    swapped = True
                    print(self.stack)   # only print when a swap happens
            print("")
            print(f"After pass {i+1}: {self.stack}")
            if not swapped:  # optimization: stop if already sorted
                break


# A generic menu function
def menu(stack_obj):
    while True:
        print("\nPlease Select any Operation:")
        print("1 - Push")
        print("2 - Pop")
        print("3 - Sort (Python built-in)")
        print("4 - Peak or Top Value")
        print("5 - isEmpty")
        print("6 - isFull")
        print("7 - Bubble Sort (Custom)")
        print("8 - Exit")

        choice = int(input("Please select option: "))
        if choice == 1:
            stack_obj.push()
        elif choice == 2:
            stack_obj.pop()
        elif choice == 3:
            stack_obj.sort()
        elif choice == 4:
            stack_obj.peak()
        elif choice == 5:
            stack_obj.isEmpty()
        elif choice == 6:
            stack_obj.isFull()
        elif choice == 7:
            stack_obj.bubble_sort()
        elif choice == 8:
            break
        else:
            print("Wrong option! Please select option ranging from 1 to 8")


# Example usage with multiple objects
obj1 = Stack(7)
obj2 = Stack(3)

print("\n--- Menu for obj1 ---")
menu(obj1)

print("\n--- Menu for obj2 ---")
menu(obj2)
```

```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(n - i - 1):  # Simplified loop range
            if arr[j] > arr[j + 1]:
                # Capture original values before swapping
                left, right = arr[j], arr[j + 1]
                arr[j], arr[j + 1] = right, left
                swapped = True
                print(f"Swap {left} and {right}: {arr}")
        print(f"Pass {i + 1}: {arr}\n")
        if not swapped:  # Early exit if no swaps
            break
    return arr

# Example usage
data = [64, 34, 25, 12, 22, 11, 90]
print(f"Unsorted list: {data}\n")
bubble_sort(data)
print(f"Sorted list: {data}")
```

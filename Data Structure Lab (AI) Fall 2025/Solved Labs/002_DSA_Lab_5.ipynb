# Task 1

```python
def insertionSort(arr):
    step=0
    cmp=0
    n=len(arr)
    for i in range(1, n):
        for j in range(i-1,-1,-1):
            cmp+=1
            if arr[j]>arr[j+1]:
                arr[j],arr[j+1]=arr[j+1],arr[j]
                step+=1
            else:
                break
    print("Compare count is: ", cmp)
    print("Swap count is: ", step)
    
    return arr
```

```python
sort_arr=[1,2,3,4,5,6]
reverse_arr=[6,5,4,3,2,1]
random_arr=[4,2,5,32,5,9]

sort=insertionSort(sort_arr)
print("Output of sorted array: ",sort,"\n")

reverse=insertionSort(reverse_arr)
print("Output of reverse sorted array: ",reverse,"\n")

random=insertionSort(random_arr)
print("Output of random array: ",random,"\n")
```

```python
# Bubble Sort
def bubble_sort(arr):
    cmp = 0
    swp = 0
    n = len(arr)
    
    for i in range(n-1):
        for j in range(0, n - i - 1):
            cmp += 1
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
            swp += 1
            
    print("Compare count is:", cmp)
    print("Swap count is:", swp)
    return arr

# Selection Sort
def selection_sort(arr):
    cmp = 0
    swp = 0
    n = len(arr)
    
    for i in range(n - 1):
        min_index = i
        for j in range(i + 1, n):
            cmp += 1
            if arr[j] < arr[min_index]:
                min_index = j
        if min_index != i:
            arr[i], arr[min_index] = arr[min_index], arr[i]
            swp += 1
            
    print("Compare count is:", cmp)
    print("Swap count is:", swp)
    return arr

# Insertion Sort
def insertion_sort(arr):
    step=0
    cmp=0
    n=len(arr)
    for i in range(1, n):
        for j in range(i-1,-1,-1):
            cmp+=1
            if arr[j]>arr[j+1]:
                arr[j],arr[j+1]=arr[j+1],arr[j]
                step+=1
            else:
                break
    print("Compare count is: ", cmp)
    print("Swap count is: ", step)
    
    return arr
```

```python
sorted_arr=[1,2,3,4,5,6]
reverse_sorted_arr=[6,5,4,3,2,1]
random_arr=[4,2,5,32,5,9]

bbl_sort_sorted=bubble_sort(sorted_arr)
print("Output of Sorted array using bubble sort: ",bbl_sort_sorted, "\n")

print("---------------------------------------------------------------------\n")

sorted_arr=[1,2,3,4,5,6]
reverse_sorted_arr=[6,5,4,3,2,1]
random_arr=[4,2,5,32,5,9]

select_sort_sorted=selection_sort(sorted_arr)
print("Output of Sorted array using selection sort: ",select_sort_sorted, "\n")

print("---------------------------------------------------------------------\n")

sorted_arr=[1,2,3,4,5,6]
reverse_sorted_arr=[6,5,4,3,2,1]
random_arr=[4,2,5,32,5,9]

insert_sort_sorted=insertion_sort(sorted_arr)
print("Output of Sorted array using insertion sort: ",insert_sort_sorted, "\n")

print("---------------------------------------------------------------------\n")

bbl_sort_reverse=bubble_sort(reverse_sorted_arr)
print("Output of reverse Sorted array using bubble sort: ",bbl_sort_reverse, "\n")

print("---------------------------------------------------------------------\n")

sorted_arr=[1,2,3,4,5,6]
reverse_sorted_arr=[6,5,4,3,2,1]
random_arr=[4,2,5,32,5,9]

select_sort_reverse=selection_sort(reverse_sorted_arr)
print("Output of reverse Sorted array using selection sort: ",select_sort_reverse, "\n")

print("---------------------------------------------------------------------\n")

sorted_arr=[1,2,3,4,5,6]
reverse_sorted_arr=[6,5,4,3,2,1]
random_arr=[4,2,5,32,5,9]

insert_sort_reverse=insertion_sort(reverse_sorted_arr)
print("Output of reverse Sorted array using insertion sort: ",insert_sort_reverse, "\n")

print("---------------------------------------------------------------------\n")

bbl_sort_random=bubble_sort(random_arr)
print("Output of random array using bubble sort: ",bbl_sort_random, "\n")

print("---------------------------------------------------------------------\n")

sorted_arr=[1,2,3,4,5,6]
reverse_sorted_arr=[6,5,4,3,2,1]
random_arr=[4,2,5,32,5,9]

select_sort_random=selection_sort(random_arr)
print("Output of random array using selection sort: ",select_sort_random, "\n")

print("---------------------------------------------------------------------\n")

sorted_arr=[1,2,3,4,5,6]
reverse_sorted_arr=[6,5,4,3,2,1]
random_arr=[4,2,5,32,5,9]

insert_sort_random=insertion_sort(random_arr)
print("Output of random array using insertion sort: ",insert_sort_random, "\n")
```

```python
#Left to Right sorting
def insertion_sort_ltr(arr):
    swp=0
    cmp=0
    n=len(arr)
    for i in range(1, n):
        for j in range(i-1,-1,-1):
            cmp+=1
            if arr[j]>arr[j+1]:
                arr[j],arr[j+1]=arr[j+1],arr[j]
                swp+=1
            else:
                break
    
    print("Compare count is: ", cmp)
    print("Swap count is: ", swp)
    
    return arr

#Right to left sorting
def insertion_sort_rtl(arr):
    swp = 0
    cmp = 0
    n = len(arr)
    for i in range(n - 2, -1, -1):
        for j in range(i + 1, n):
            cmp += 1
            if arr[j] < arr[j - 1]:
                arr[j], arr[j - 1] = arr[j - 1], arr[j]
                swp += 1
            else:
                break

    print("Compare count is: ", cmp)
    print("Swap count is: ", swp)

    return arr
```

```python
random_arr=[4,2,5,32,5,9]

out3=insertion_sort_rtl(random_arr)
print("Output of random array using insertion sort: ",out3, "\n")

random_arr=[4,2,5,32,5,9]

out3=insertion_sort_ltr(random_arr)
print("Output of random array using insertion sort: ",out3, "\n")
```

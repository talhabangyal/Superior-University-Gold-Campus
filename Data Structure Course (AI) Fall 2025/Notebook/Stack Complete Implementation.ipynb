```python
import tkinter as tk
from tkinter import messagebox, simpledialog


class Stack:
    def __init__(self, total_value):
        self.stack = []
        self.limit = total_value

    def push(self, value):
        if len(self.stack) == self.limit:
            return "Stack is already Filled"
        else:
            self.stack.append(value)
            return f"Pushed {value}"

    def pop(self):
        if not self.stack:
            return "Stack is already Empty"
        else:
            j = self.stack.pop()
            return f"Popped {j}"

    def peak(self):
        if not self.stack:
            return "No top value exist"
        else:
            return f"Top Value: {self.stack[-1]}"

    def isFull(self):
        if self.limit == len(self.stack):
            return "Stack is Full"
        else:
            return "Stack is Not Full"

    def isEmpty(self):
        if not self.stack:
            return "Stack is Empty"
        else:
            return "Stack is Not Empty"

    def sort(self):
        self.stack.sort()
        asc = self.stack.copy()
        self.stack.sort(reverse=True)
        desc = self.stack.copy()
        return f"Ascending: {asc}\nDescending: {desc}"


class StackGUI:
    def __init__(self, root, stack_obj):
        self.stack_obj = stack_obj
        self.root = root
        self.root.title("Stack Operations GUI")

        # Display Stack
        self.stack_label = tk.Label(root, text="Stack: []", font=("Arial", 14))
        self.stack_label.pack(pady=10)

        # Buttons
        tk.Button(root, text="Push", width=15, command=self.push_value).pack(pady=5)
        tk.Button(root, text="Pop", width=15, command=self.pop_value).pack(pady=5)
        tk.Button(root, text="Sort", width=15, command=self.sort_stack).pack(pady=5)
        tk.Button(root, text="Peak", width=15, command=self.peak_value).pack(pady=5)
        tk.Button(root, text="isEmpty", width=15, command=self.check_empty).pack(pady=5)
        tk.Button(root, text="isFull", width=15, command=self.check_full).pack(pady=5)

    def update_stack_display(self):
        self.stack_label.config(text=f"Stack: {self.stack_obj.stack}")

    def push_value(self):
        if len(self.stack_obj.stack) < self.stack_obj.limit:
            value = simpledialog.askstring("Input", "Enter value to push:")
            if value:
                msg = self.stack_obj.push(value)
                messagebox.showinfo("Push", msg)
        else:
            messagebox.showwarning("Warning", "Stack is already filled")
        self.update_stack_display()

    def pop_value(self):
        msg = self.stack_obj.pop()
        messagebox.showinfo("Pop", msg)
        self.update_stack_display()

    def peak_value(self):
        msg = self.stack_obj.peak()
        messagebox.showinfo("Peak", msg)

    def check_empty(self):
        msg = self.stack_obj.isEmpty()
        messagebox.showinfo("isEmpty", msg)

    def check_full(self):
        msg = self.stack_obj.isFull()
        messagebox.showinfo("isFull", msg)

    def sort_stack(self):
        msg = self.stack_obj.sort()
        messagebox.showinfo("Sort", msg)
        self.update_stack_display()


# Main Program
if __name__ == "__main__":
    root = tk.Tk()
    obj1 = Stack(5)  # Create a stack with size 5
    gui = StackGUI(root, obj1)
    root.mainloop()
```

```python
class Stack:
    def __init__(self, total_value):
        self.stack = []
        self.limit = total_value
        print(f"\n✅ New Stack created with limit {self.limit}")

    def push(self):
        if len(self.stack) == self.limit:
            print("⚠️ Stack is already full! Cannot push more values.")
        else:
            while True:
                try:
                    e = int(input("Enter an integer value to push: "))
                    self.stack.append(e)
                    print(f"✅ Value {e} pushed! Current Stack: {self.stack}")
                    break
                except ValueError:
                    print("❌ Invalid input! Please enter an INTEGER only.")

    def pop(self):
        if not self.stack:
            print("⚠️ Stack is already empty!")
        else:
            removed = self.stack.pop()
            print(f"✅ Removed value: {removed}")
            print("Current Stack:", self.stack)

    def peek(self):
        if not self.stack:
            print("⚠️ No top value (Stack is empty).")
        else:
            print(f"🔝 Top Value: {self.stack[-1]}")

    def isFull(self):
        if len(self.stack) == self.limit:
            print("✅ Stack is FULL")
        else:
            print(f"ℹ️ Stack is NOT Full. Current size: {len(self.stack)}/{self.limit}")

    def isEmpty(self):
        if not self.stack:
            print("✅ Stack is EMPTY")
        else:
            print("ℹ️ Stack is NOT Empty. Current Stack:", self.stack)

    def sort(self):
        if not self.stack:
            print("⚠️ Cannot sort an empty stack.")
            return
        ascending = sorted(self.stack)
        descending = sorted(self.stack, reverse=True)
        print("⬆️ Ascending Order:", ascending)
        print("⬇️ Descending Order:", descending)


# Menu function
def menu(stack_obj):
    while True:
        print("\n📌 Select an Operation:")
        print("1 - Push")
        print("2 - Pop")
        print("3 - Sort")
        print("4 - Peek (Top Value)")
        print("5 - Check if Empty")
        print("6 - Check if Full")
        print("7 - Exit")

        try:
            choice = int(input("👉 Enter your choice (1-7): "))
        except ValueError:
            print("❌ Invalid input! Please enter a number between 1-7.")
            continue

        if choice == 1:
            stack_obj.push()
        elif choice == 2:
            stack_obj.pop()
        elif choice == 3:
            stack_obj.sort()
        elif choice == 4:
            stack_obj.peek()
        elif choice == 5:
            stack_obj.isEmpty()
        elif choice == 6:
            stack_obj.isFull()
        elif choice == 7:
            print("👋 Exiting menu...")
            break
        else:
            print("⚠️ Wrong option! Please select between 1 and 7.")


# Example usage
obj1 = Stack(4)
obj2 = Stack(3)

print("\n--- Menu for Stack obj1 ---")
menu(obj1)

print("\n--- Menu for Stack obj2 ---")
menu(obj2)
```
